```python
# Importing necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Generating synthetic data for linear regression
bias = 250
X,y, coef = make_regression(bias=bias, n_features=1, noise=10, coef=True)
```

```python
# Data shapes and coefficients
X.shape, y.shape, coef
```

```python
# Plotting the Data
plt.scatter(X, y)
plt.title("Scatter plot of the generated data")
plt.show()
```

```python
# Line Equation
Y = coef * X + bias
```

```python
# Plotting the Regression Line
plt.scatter(X, y)
plt.plot(X,Y, color='red', label='Regression Line')
plt.title("Scatter plot with Regression Line")
plt.legend()
plt.show()
```

```python
# Training the Linear Regression Model
model = LinearRegression()
y_predict = model.fit(X, y)
```

```python
# Model Coefficients and Intercept
model.coef_, model.intercept_
```

```python
# Plotting the Regression Line and Fitted Line
plt.scatter(X, y)
plt.plot(X, Y, color='red', label='Regression Line')
plt.plot(X, model.predict(X), color='green', label='Fitted Line')
plt.title("Scatter plot with Regression Line and Fitted Line")
plt.legend()
plt.show()
```

```python
# Model Evaluation with R^2 Score
accuracy = model.score(X, y)
print(f"Model Accuracy (R^2 score): {accuracy:.4f}")
```

```python
# Model Evaluation with equation line 
accuracy = model.score(X, Y)
print(f"Model Accuracy (R^2 score): {accuracy:.4f}")
```
